# CCM and ECM calculation:
**Task:** 
1. Get the nearest symmetric structure from https://csm.ouproj.org.il/molecule (automatically)
2. Extract the chiral and achiral structures from the downloaded zip
4. Calculate and print CCMs and ECM

```python
# creating the path (PYTHONPATH) to our module.
# assuming that our 'pyECM' directory is out ('..') of our current directory 
import os
import sys
module_path = os.path.abspath(os.path.join('..'))

if module_path not in sys.path:
    sys.path.append(module_path)
```

```python
import pathlib
from pyECM.data_management import obtain_nearest_structure
from pyECM.molecule_class import molecula


MOL='AP1'
input_dir = '../pyECM/data/import'
output_dir = '../pyECM/data/export'

obtain_nearest_structure(MOL,input_dir,output_path=output_dir)
```

```python
chiral_file = output_dir + '/'+ MOL + '_chiral'
basis='sto-3g'

# Import chiral and achiral structure and export files to "export" folder.
mymolecule = molecula(XYZ_file = chiral_file + '.xyz', XYZ_achiral_file = output_dir + '/' + MOL + '_achiral.xyz')
mymolecule.export_xyz(folder=output_dir+ '/', prefix_name=MOL+'_chiral', DIRAC = False, achiral= output_dir + '/' + MOL + '_achiral.xyz')

# Calculate CCM
mymolecule.CCM()

#Define options for the Gaussian Type Orbitals (generated by pySCF) and the WF calculation.
GTO={'basis' : basis, 'charge' : 0, 'spin' : 0}
WF_method = {'fourcomp': False, 'debug' : 0, 'X2C' : True}

# Get the WF with the pySCF code
mymolecule.pySCF_WF(name=output_dir+ '/' + MOL + '_chiral', gto_dict=GTO, method_dict=WF_method)

#Calculate ECM
method = {'fourcomp': False, 'debug' : 0, 'X2C' : True}
mymolecule.ECM(name=output_dir+ '/' + MOL + '_chiral', method_dict=method)

#Print section
print("CCMs: ",mymolecule.CCM1, mymolecule.CCM2)
print("Non-rel and x2c ECMs: ", mymolecule.ECM_NR, mymolecule.ECM_X2C)
```
